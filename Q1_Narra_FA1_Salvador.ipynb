```python
import math

# This are the variables needed for distance formula.
x1 = float(input("What is the value of x1: "))
y1 = float(input("What is the value of y1: "))
x2 = float(input("What is the value of x2: "))
y2 = float(input("What is the value of y2: "))

# This code does the distance formula using the variables gained.
# It rounds to the 2nd decimal point.
distance = round(math.sqrt(pow(x2-x1,2) + pow(y2-y1,2)), 2)

# This prints the final output.
print(distance)
```
