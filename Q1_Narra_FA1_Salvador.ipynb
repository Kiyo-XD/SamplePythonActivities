```python
import math
x1 = float(str(input("What is the value of x1: ")))
y1 = float(str(input("What is the value of y1: ")))
x2 = float(str(input("What is the value of x2: ")))
y2 = float(str(input("What is the value of y2: ")))
d = math.sqrt(pow(x2-x1,2) + pow(y2-y1,2))
print(d)
```
